```python
import numpy as np

data = np.array([45, 78, 23, 90, 56, 12, 67, 50, 89, 34])
print("Original data :", data)

# Boolean array
mask = data > 50
print("Boolean array (data > 50):", mask)

# Boolean indexing
print("Above 50      :", data[data > 50])
print("Below 50      :", data[data < 50])
print("Equal to 50   :", data[data == 50])
print("50 or more    :", data[data >= 50])

# Different thresholds
for t in [30, 50, 70]:
    print(f"Greater than {t}: {data[data > t]}")

# Combined condition
print("Between 40 and 80:", data[(data >= 40) & (data <= 80)])

# Count
print("Count above 50:", np.sum(data > 50))
```
